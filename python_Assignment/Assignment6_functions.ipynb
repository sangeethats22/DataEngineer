## Assignment 12: Check Leap Year

Write a function named `is_leap_year()`.

### Requirements

- Accept a year as an integer.
- Return `True` when it is a leap year.
- Otherwise, return `False`.
- Print an appropriate message.

```python
def is_leap_year(year):
    # Leap year rules:
    # 1. Divisible by 4 → leap year
    # 2. if divisible by 100 → not leap year
    # 3. if divisible by 400 → leap year
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return True
    else:
        return False

year = int(input("Enter a year: "))

if is_leap_year(year):
    print(year, "is a Leap Year.")
else:
    print(year, "is NOT a Leap Year.")
```

## Assignment 13: Factorial Using a Function

Write a function named `calculate_factorial()`.

### Requirements

- Accept a non-negative integer.
- Calculate its factorial using a loop.
- Do not use `math.factorial()`.
- Return the factorial.
- Handle negative input.

```python
def calculate_factorial(n):
    if n < 0:
        return "Error! Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        fact = 1
        for i in range(2, n + 1):
            fact *= i
        return fact

num = int(input("Enter a non-negative integer: "))

result = calculate_factorial(num)
print("Factorial of", num, "is:", result)
```

- Write another function that reverses it using a loop.

```python
# Method 1: Using slicing
def reverse_string(text):
    return text[::-1]

# Method 2: Using a loop
def reverse_string_loop(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text
    return reversed_text

# Accept a string from the user
user_input = input("Enter a string: ")

# Call both functions
print("Reversed using slicing:", reverse_string(user_input))
print("Reversed using loop:", reverse_string_loop(user_input))
```

## Assignment 17: Palindrome Checker

Write a function named `is_palindrome()`.

### Requirements

- Accept a string.
- Ignore spaces and letter case.
- Return `True` when the string is a palindrome.
- Print a meaningful result.

```python
def is_palindrome(text):
    # Remove spaces and convert to lowercase
    cleaned_text = text.replace(" ", "").lower()

    # Check if the cleaned text equals its reverse
    return cleaned_text == cleaned_text[::-1]

user_input = input("Enter a string: ")

if is_palindrome(user_input):
    print(f'"{user_input}" is a Palindrome.')
else:
    print(f'"{user_input}" is NOT a Palindrome.')
```

## Assignment 19: Find Maximum in a List

Write a function named `find_list_maximum()`.

### Requirements

- Accept a list of numbers.
- Do not use `max()`.
- Return the largest list element.
- Handle an empty list by returning an appropriate message.


```python
def find_list_maximum(numbers):
    if not numbers:  # check for empty list
        return "Error! The list is empty."

    maximum = numbers[0]  # assume first element is max
    for num in numbers[1:]:
        if num > maximum:
            maximum = num
    return maximum

user_input = input("Enter numbers separated by spaces: ")
numbers_list = [float(x) for x in user_input.split()]

result = find_list_maximum(numbers_list)
print("The largest element is:", result)
```

Write a function named `count_number_types()`.

### Requirements

- Accept a list of integers.
- Count positive numbers, negative numbers, and zeros.
- Return all three counts.
- Use tuple unpacking to receive the returned values.

```python
def count_number_types(numbers):
    positives = 0
    negatives = 0
    zeros = 0

    for num in numbers:
        if num > 0:
            positives += 1
        elif num < 0:
            negatives += 1
        else:
            zeros += 1

    return positives, negatives, zeros  # return as a tuple

user_input = input("Enter integers separated by spaces: ")
numbers_list = [int(x) for x in user_input.split()]

# Use tuple unpacking to receive the returned values
pos_count, neg_count, zero_count = count_number_types(numbers_list)

print("Positive numbers:", pos_count)
print("Negative numbers:", neg_count)
print("Zeros:", zero_count)
```

## Assignment 33: Function Returning a List

Write a function named `get_even_numbers()`.

### Requirements

- Accept a starting number and ending number.
- Return a list containing all even numbers in the range.
- Validate that the starting number is smaller than the ending number.

```python
def get_even_numbers(start, end):
    # Validate that start is smaller than end
    if start >= end:
        print("Error: Starting number must be smaller than ending number.")
        return []

    even_list = []
    for num in range(start, end + 1):
        if num % 2 == 0:
            even_list.append(num)

    return even_list


print("Even numbers from 1 to 10:", get_even_numbers(1, 10))
print("Even numbers from 20 to 30:", get_even_numbers(20, 30))

# Invalid case (start >= end)
print("Invalid case:", get_even_numbers(15, 10))
```

## Assignment 34: Prime Number Function

Write a function named `is_prime()`.

### Requirements

- Accept one integer.
- Return `True` when the number is prime.
- Return `False` otherwise.
- Use the function to print all prime numbers from 1 to 100.

```python
def is_prime(n):
    # Handle edge cases
    if n <= 1:
        return False
    if n == 2:
        return True

    # Check divisibility up to sqrt(n)
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100:")
for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

```python
def generate_fibonacci(n_terms):
    # Handle invalid cases
    if n_terms <= 0:
        print("Error: Number of terms must be positive.")
        return []

    # Special cases
    if n_terms == 1:
        return [0]
    elif n_terms == 2:
        return [0, 1]

    # Generate Fibonacci series iteratively (no recursion)
    series = [0, 1]
    for i in range(2, n_terms):
        next_value = series[-1] + series[-2]
        series.append(next_value)

    return series

print("Fibonacci (0 terms):", generate_fibonacci(0))
print("Fibonacci (1 term):", generate_fibonacci(1))
print("Fibonacci (10 terms):", generate_fibonacci(10))
```

## Assignment 92: Closure-Based Bank Balance

Write a function named `create_bank_account()`.

### Requirements

- Store the balance in the outer function.
- Define inner functions for:
  - Deposit
  - Withdrawal
  - Balance enquiry
- Use `nonlocal` to update the balance.
- Reject withdrawals greater than the available balance.
- Return all three inner functions.


```python
def create_bank_account(initial_balance=0):
    balance = initial_balance  # Outer variable to store account balance

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            return f"Deposited {amount}. New balance: {balance}"
        else:
            return "Deposit amount must be positive."

    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            return "Withdrawal amount must be positive."
        elif amount > balance:
            return f"Insufficient funds! Current balance: {balance}"
        else:
            balance -= amount
            return f"Withdrew {amount}. New balance: {balance}"

    def enquiry():
        return f"Current balance: {balance}"

    # Return all three inner functions
    return deposit, withdraw, enquiry


deposit, withdraw, enquiry = create_bank_account(1000)

print(enquiry())          # Current balance: 1000
print(deposit(500))       # Deposited 500. New balance: 1500
print(withdraw(200))      # Withdrew 200. New balance: 1300
print(withdraw(2000))     # Insufficient funds! Current balance: 1300
print(enquiry())          # Current balance: 1300
```

## Assignment 93: Function Decorator

### Requirements

- The decorator should print a message before the original function executes.
- It should print another message after execution.
- Apply it to an addition function.
- Preserve the original function's return value.

```python
def execution_logger(func):
    def wrapper(*args, **kwargs):
        print(f"Starting execution of '{func.__name__}'...")
        result = func(*args, **kwargs)   # Call the original function
        print(f"Finished execution of '{func.__name__}'.")
        return result  # Preserve the return value
    return wrapper


@execution_logger
def add(a, b):
    return a + b


x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

result = add(x, y)
print("Result:", result)
```

```python
from functools import reduce

# -------------------------------
# Global Variable for LEGB Demo
# -------------------------------
company_name = "Edukron"

# -------------------------------
# Sample Employee Data
# -------------------------------
employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2}
]

# -------------------------------
# Part 1: Display Employees
# -------------------------------
def display_employees(emp_list):
    if not emp_list:
        print("No employees found.")
        return
    for emp in emp_list:
        print(f"ID: {emp['id']}, Name: {emp['name']}, Dept: {emp['department']}, "
              f"Salary: {emp['salary']}, Performance: {emp['performance_score']}, Experience: {emp['experience']}")

# -------------------------------
# Part 2: Extract Employee Names
# -------------------------------
def extract_names(emp_list):
    return list(map(lambda e: e["name"], emp_list))

# -------------------------------
# Part 3: Filter High Performers
# -------------------------------
def filter_high_performers(emp_list):
    return list(filter(lambda e: e["performance_score"] >= 80, emp_list))

# -------------------------------
# Part 4: Filter by Department
# -------------------------------
def filter_by_department(emp_list, dept):
    return list(filter(lambda e: e["department"].lower() == dept.lower(), emp_list))

# -------------------------------
# Part 5: Salary Increase
# -------------------------------
def calculate_revised_salaries(emp_list):
    def revised(emp):
        if emp["performance_score"] >= 90:
            increase = emp["salary"] * 0.15
        elif emp["performance_score"] >= 80:
            increase = emp["salary"] * 0.10
        elif emp["performance_score"] >= 70:
            increase = emp["salary"] * 0.05
        else:
            increase = 0
        return {"name": emp["name"], "old_salary": emp["salary"], "new_salary": emp["salary"] + increase}
    return list(map(revised, emp_list))

# -------------------------------
# Part 6: Total Salary Expense
# -------------------------------
def calculate_salary_expense(emp_list, revised_list):
    current_total = reduce(lambda acc, e: acc + e["salary"], emp_list, 0)
    revised_total = reduce(lambda acc, e: acc + e["new_salary"], revised_list, 0)
    additional_expense = revised_total - current_total
    return current_total, revised_total, additional_expense

# -------------------------------
# Part 7: Highest-Paid Employee
# -------------------------------
def highest_paid(emp_list):
    return reduce(lambda acc, e: e if e["salary"] > acc["salary"] else acc, emp_list)

# -------------------------------
# Part 8: Sort Employees
# -------------------------------
def sort_employees(emp_list, key):
    return sorted(emp_list, key=lambda e: e[key])

# -------------------------------
# Part 9: Closure-Based Bonus Calculator
# -------------------------------
def create_bonus_calculator(bonus_percent):
    def bonus(salary):
        return salary * bonus_percent / 100
    return bonus

bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

# -------------------------------
# Part 10 & 11: Nested Report Generator + LEGB Demo
# -------------------------------
def generate_department_report(emp_list, dept):
    department = dept  # Enclosing variable

    def num_employees():
        return len(filter_by_department(emp_list, department))

    def total_salary():
        return reduce(lambda acc, e: acc + e["salary"], filter_by_department(emp_list, department), 0)

    def avg_salary():
        dept_emps = filter_by_department(emp_list, department)
        return total_salary() / len(dept_emps) if dept_emps else 0

    def avg_performance():
        dept_emps = filter_by_department(emp_list, department)
        return reduce(lambda acc, e: acc + e["performance_score"], dept_emps, 0) / len(dept_emps) if dept_emps else 0

    # Local variable
    report_title = f"{company_name} - {department} Department Report"

    return {
        "title": report_title,
        "num_employees": num_employees(),
        "total_salary": total_salary(),
        "avg_salary": avg_salary(),
        "avg_performance": avg_performance()
    }

"""
LEGB Explanation:
- Global: company_name = "Edukron"
- Enclosing: department variable inside generate_department_report()
- Local: report_title inside generate_department_report()
- Built-in: Python built-ins like len(), reduce(), print()
"""

# -------------------------------
# Part 12: Menu-Driven Program
# -------------------------------
def menu():
    while True:
        print("\n--- Employee Performance Analysis System ---")
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_employees(employees)
        elif choice == "2":
            print("Employee Names:", extract_names(employees))
        elif choice == "3":
            display_employees(filter_high_performers(employees))
        elif choice == "4":
            dept = input("Enter department name: ")
            result = filter_by_department(employees, dept)
            if result:
                display_employees(result)
            else:
                print("No employees found in that department.")
        elif choice == "5":
            revised = calculate_revised_salaries(employees)
            for r in revised:
                print(f"{r['name']} - Old: {r['old_salary']}, New: {r['new_salary']}")
        elif choice == "6":
            revised = calculate_revised_salaries(employees)
            current, revised_total, extra = calculate_salary_expense(employees, revised)
            print(f"Current Total: {current}, Revised Total: {revised_total}, Additional Expense: {extra}")
        elif choice == "7":
            hp = highest_paid(employees)
            print(f"Highest Paid: {hp['name']} with salary {hp['salary']}")
        elif choice == "8":
            key = input("Sort by (name/salary/performance_score/experience): ")
            if key in ["name", "salary", "performance_score", "experience"]:
                display_employees(sort_employees(employees, key))
            else:
                print("Invalid sort key.")
        elif choice == "9":
            dept = input("Enter department name: ")
            report = generate_department_report(employees, dept)
            print(report)
        elif choice == "10":
            print("Exiting program...")
            break
        else:
            print("Invalid choice. Please try again.")

# Uncomment to run the menu
menu()
```
