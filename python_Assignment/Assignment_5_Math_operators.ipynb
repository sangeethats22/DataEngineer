### Bitwise Operators

```python
# 17. Find the bitwise AND, OR, XOR of two numbers.
a = 16
b = 25
print("Bitwise AND:", a & b)
print("Bitwise OR:", a | b)
print("Bitwise XOR:", a ^ b)

# 18. Perform left shift and right shift operations.
a = 10
print("Left shift:", a << 2)
print("Right shift:", a >> 2)

# 19. Check whether a number is even using a bitwise operator.
num = 12
print("Even" if num & 1 == 0 else "Odd")

# 20. Swap two numbers using the XOR operator.
a = 10
b = 20
print("Before swapping: a =", a, "b =", b)
a = a ^ b
b = a ^ b
a = a ^ b
print("After swapping: a =", a, "b =", b)

# 21. Count how many bits are set to 1 in a number using bitwise operations.
num = 12
count = 0
while num > 0:
    count += num & 1
    num >>= 1
print("Number of set bits:", count)
```

### Membership Operators

```python
# 25. Check whether a character exists in a string.
word = "samsung"
print('z' in word)

# 26. Check whether a number exists in a list
numbers = [1, 2, 3, 4, 5]
num = 3
print(num in numbers)

# 27. Count how many vowels in a string belong to a predefined vowel list using `in`.
vowels = ['a', 'e', 'i', 'o', 'u']
word = "samsung"
count = 0
for char in word:
    if char in vowels:
        count += 1
print("Number of vowels:", count)
```
